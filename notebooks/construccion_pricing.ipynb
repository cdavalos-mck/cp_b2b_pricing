```python
margen_df = catalog.load("second_stage_predictions")
detalle_estaciones = catalog.load("detalle_estaciones")
mst_transactions_channel = catalog.load("mst_transactions_channel")
```

```python
detalle_estaciones = detalle_estaciones.with_columns(codigo_pais=pl.lit("CL"))
```

```python
margen_df = margen_df.select("customer_id", "corrected_predicted_value").rename(
    {
        "corrected_predicted_value": "margen_estimado",
    }
)
```

```python
def compute_customer_prices_by_level(
    margen_df: pl.DataFrame,
    detalle_estaciones: pl.DataFrame,
    mst_transactions_channel: pl.DataFrame,
    level: str = "codigo_comuna",  # "codigo_estacion", "codigo_comuna", "codigo_region", or "codigo_pais"
) -> pl.DataFrame:
    if level not in {
        "codigo_estacion",
        "codigo_comuna",
        "codigo_localidad",
        "codigo_region",
        "codigo_pais",
    }:
        raise ValueError(f"Invalid aggregation level: {level}")

    # If codigo_pais, we fake a column to unify it
    if level == "codigo_pais":
        detalle_estaciones = detalle_estaciones.with_columns(
            codigo_pais=pl.lit("CL")  # Example: 'CL' for Chile
        )
        level_col = "codigo_pais"
    else:
        level_col = level

    # Base reference of stations
    ref_estaciones = (
        detalle_estaciones.select(
            "codigo_estacion", "codigo_localidad", "codigo_comuna", "codigo_region"
        )
        .unique()
        .with_columns(codigo_pais=pl.lit("CL"))
    )

    # Station info with cost
    estacion_df = detalle_estaciones.select(
        "codigo_estacion",
        "codigo_localidad",
        "codigo_comuna",
        "codigo_region",
        "costo_total",
        "codigo_pais",
    ).unique()

    # Cross join with margins (customer × station)
    client_station_prices = estacion_df.join(margen_df, how="cross")

    # Price = cost + margin
    client_station_df = client_station_prices.with_columns(
        precio_estacion=pl.col("costo_total") + pl.col("margen_estimado")
    ).select(
        "customer_id",
        "codigo_estacion",
        "codigo_localidad",
        "codigo_comuna",
        "codigo_region",
        "precio_estacion",
        "codigo_pais",
    )

    # Historical volumes
    trx_df = (
        mst_transactions_channel.group_by("customer_id", "estacion")
        .agg(volumen_total=pl.col("volumen").sum())
        .rename({"estacion": "codigo_estacion"})
        .join(ref_estaciones, on="codigo_estacion")
    )

    historic_df = trx_df.group_by("customer_id", "codigo_estacion", level_col).agg(
        volumen_total=pl.col("volumen_total").sum()
    )

    # Simple average per customer × level
    avg_df = client_station_df.group_by("customer_id", level_col).agg(
        precio_promedio_simple=pl.col("precio_estacion").mean()
    )

    # Combine all into one table
    client_level_df = client_station_df.join(
        historic_df,
        on=["customer_id", "codigo_estacion", level_col],
        how="left",
    ).with_columns(
        volumen_total=pl.col("volumen_total").fill_null(0),
        revenue=pl.col("precio_estacion") * pl.col("volumen_total"),
    )

    # Weighted average per customer × level
    client_level_df = (
        client_level_df.group_by("customer_id", level_col)
        .agg(
            revenue_total=pl.col("revenue").sum(),
            volumen_total=pl.col("volumen_total").sum(),
        )
        .with_columns(
            precio_promedio_ponderado=(
                pl.col("revenue_total") / pl.col("volumen_total")
            ).round(2)
        )
    )

    # Final price per customer × level
    final_df = (
        client_level_df.join(avg_df, on=["customer_id", level_col], how="left")
        .with_columns(
            precio_final=pl.when(pl.col("precio_promedio_ponderado").is_nan())
            .then(pl.col("precio_promedio_simple"))
            .otherwise(pl.col("precio_promedio_ponderado"))
        )
        .select("customer_id", level_col, "precio_final")
    )

    final_df = (
        final_df.with_columns(
            max_precio_comuna=pl.col("precio_final").max().over(level_col)
        )
        .with_columns(
            descuento_comuna=pl.col("max_precio_comuna") - pl.col("precio_final")
        )
        .select("customer_id", level_col, "precio_final", "descuento_comuna")
    )

    return final_df
```

```python
precio_comuna_df = compute_customer_prices_by_level(
    margen_df, detalle_estaciones, mst_transactions_channel, level="codigo_comuna"
)

precio_localidad_df = compute_customer_prices_by_level(
    margen_df, detalle_estaciones, mst_transactions_channel, level="codigo_localidad"
)

precio_region_df = compute_customer_prices_by_level(
    margen_df, detalle_estaciones, mst_transactions_channel, level="codigo_region"
)

precio_pais_df = compute_customer_prices_by_level(
    margen_df, detalle_estaciones, mst_transactions_channel, level="codigo_pais"
)
```
