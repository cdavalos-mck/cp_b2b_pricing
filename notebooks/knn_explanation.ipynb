```python
from dash import Input, Output, dash_table

# Load your data
explanation = explanation
neighbors_df = closest_customers_detail

# Define percentage share features
perc_features = [
    "share_customer_region_14",
    "share_customer_region_11",
    "share_customer_region_5",
    "share_customer_region_9",
    "share_customer_region_4",
    "share_customer_region_8",
    "share_customer_region_12",
    "share_customer_region_15",
    "share_customer_region_16",
    "share_customer_region_6",
    "share_customer_region_7",
    "share_customer_region_10",
    "share_customer_region_1",
    "share_customer_region_3",
    "share_customer_region_13",
    "share_customer_region_2",
]

# App
app = Dash(__name__)

# Initial column options for the column selector
all_columns = [
    {"label": col, "value": col}
    for col in neighbors_df.columns
    if col != "source_customer_id"
]

app.layout = html.Div(
    [
        html.H2("Customer Feature Breakdown"),
        dcc.Dropdown(
            id="customer-selector",
            options=[
                {"label": cid, "value": cid}
                for cid in explanation["customer_id"].unique()
            ],
            value=explanation["customer_id"].unique()[0],
            style={"width": "400px", "margin-bottom": "20px"},
        ),
        dcc.Graph(id="polar-subplots"),
        html.H3("Neighbor Detail Table"),
        html.Div(
            [
                html.Label("Select columns to show:"),
                dcc.Dropdown(
                    id="column-selector",
                    options=all_columns,
                    value=[
                        all_columns[0]["value"]
                    ],  # Start with the first column selected
                    multi=True,
                    style={"width": "100%", "margin-bottom": "10px"},
                ),
            ]
        ),
        dash_table.DataTable(
            id="neighbors-table",
            columns=[],
            data=[],
            page_size=10,
            style_table={"overflowX": "auto"},
            style_cell={
                "textAlign": "left",
                "minWidth": "120px",
                "maxWidth": "300px",
                "whiteSpace": "normal",
            },
            style_header={"backgroundColor": "lightgrey", "fontWeight": "bold"},
        ),
    ]
)


@app.callback(Output("polar-subplots", "figure"), Input("customer-selector", "value"))
def update_chart(customer_id):
    test = explanation[explanation["customer_id"] == customer_id]

    perc_features_df = test[test["feature"].isin(perc_features)]
    non_perc_features_df = test[~test["feature"].isin(perc_features)]

    fig = make_subplots(
        rows=1,
        cols=2,
        specs=[[{"type": "polar"}, {"type": "polar"}]],
        subplot_titles=("Percentage Share Features", "Other Scaled Features"),
    )

    fig.add_trace(
        go.Scatterpolar(
            r=perc_features_df["customer_value"],
            theta=perc_features_df["feature"],
            fill="toself",
            line=dict(color="red", width=3),
            name="Customer Value (Perc)",
        ),
        row=1,
        col=1,
    )

    fig.add_trace(
        go.Scatterpolar(
            r=perc_features_df["avg_neighbor_value"],
            theta=perc_features_df["feature"],
            fill="toself",
            line=dict(color="blue", width=3),
            opacity=0.6,
            name="Avg Neighbor (Perc)",
        ),
        row=1,
        col=1,
    )

    fig.add_trace(
        go.Scatterpolar(
            r=non_perc_features_df["scaled_customer_value"],
            theta=non_perc_features_df["feature"],
            fill="toself",
            line=dict(color="red", width=3),
            name="Customer Value (Scaled)",
        ),
        row=1,
        col=2,
    )

    fig.add_trace(
        go.Scatterpolar(
            r=non_perc_features_df["scaled_avg_neighbor_value"],
            theta=non_perc_features_df["feature"],
            fill="toself",
            line=dict(color="blue", width=3),
            opacity=0.6,
            name="Avg Neighbor (Scaled)",
        ),
        row=1,
        col=2,
    )

    fig.update_layout(
        title_text=f"Feature Breakdown for Customer {customer_id}",
        height=650,
        showlegend=True,
    )

    return fig


@app.callback(
    Output("neighbors-table", "columns"),
    Output("neighbors-table", "data"),
    Input("customer-selector", "value"),
    Input("column-selector", "value"),
)
def update_neighbors_table(customer_id, selected_columns):
    if not selected_columns:
        return [], []

    # Ensure 'source_customer_id' is always shown
    fixed_column = "source_customer_id"
    columns_to_show = [fixed_column] + [
        col for col in selected_columns if col != fixed_column
    ]

    # Filter neighbors for the selected customer_id
    filtered = neighbors_df[neighbors_df[fixed_column] == customer_id]

    # Format the data for display
    formatted_data = []
    for _, row in filtered.iterrows():
        formatted_row = {}
        for col in columns_to_show:
            value = row.get(col, "")
            if pd.isna(value):
                formatted_row[col] = ""
            elif col in perc_features:
                formatted_row[col] = (
                    f"{value:.2%}" if isinstance(value, (float, int)) else value
                )
            elif isinstance(value, (float, int)):
                formatted_row[col] = f"{value:.2f}"
            else:
                formatted_row[col] = value
        formatted_data.append(formatted_row)

    # Define columns for dash table
    columns = [{"name": "Customer ID", "id": fixed_column}] + [
        {"name": col, "id": col} for col in selected_columns if col != fixed_column
    ]

    return columns, formatted_data


if __name__ == "__main__":
    app.run(debug=True)
```
